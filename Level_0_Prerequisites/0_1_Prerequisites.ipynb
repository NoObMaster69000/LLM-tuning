### 0.1 What is an LLM?

A Large Language Model (LLM) is a type of artificial intelligence (AI) model that is designed to understand and generate human language. Key characteristics include:

- **Transformer-based Architecture**: Most modern LLMs (like GPT, Llama, Mistral) are built on the Transformer architecture, which uses attention mechanisms to process sequential data.
- **Pre-trained**: They are pre-trained on massive amounts of text and code from the internet. This phase is computationally expensive and gives the model its general knowledge.
- **Zero-shot/Few-shot Inference**: Thanks to pre-training, LLMs can often perform tasks they weren't explicitly trained on (zero-shot) or with just a few examples provided in the prompt (few-shot).

### 0.2 Core Concepts

- **Pre-training vs. Fine-tuning**: 
  - **Pre-training**: The initial, resource-intensive training on a massive, general dataset. This is usually done by large AI labs.
  - **Fine-tuning**: The process of taking a pre-trained model and further training it on a smaller, task-specific dataset. This is what this roadmap is all about!

- **Prompting vs. Parameter Tuning**:
  - **Prompting**: Guiding the model's output by carefully crafting the input text (the prompt). No model weights are changed.
  - **Parameter Tuning**: Actually updating the model's weights (parameters) through fine-tuning.

- **Key Inference Parameters**:
  - **Context Window**: The maximum number of tokens the model can consider at once.
  - **Tokens**: The basic units of text the model processes (roughly equivalent to 4 characters).
  - **Temperature**: Controls randomness. Higher values (e.g., 0.8) make output more random, while lower values (e.g., 0.2) make it more deterministic.
  - **Top-p (Nucleus Sampling)**: A method for sampling where the model considers only the most probable tokens with a cumulative probability of `p`.
  - **Max Length**: The maximum number of tokens to generate.

### 0.4 Hardware

- **GPU (Graphics Processing Unit)**: A powerful GPU is essential for fine-tuning. NVIDIA GPUs with CUDA support are the standard.
- **VRAM (Video RAM)**: This is the memory on your GPU. It's often the biggest bottleneck.
  - **16GB+ VRAM**: Required for full fine-tuning of smaller models.
  - **8GB+ VRAM**: Can be sufficient for parameter-efficient methods like QLoRA, which we'll cover later.
